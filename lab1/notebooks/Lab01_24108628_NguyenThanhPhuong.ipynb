1. Agent là gì?
- Agent là tác nhân tương tác với environment, quan sát trạng thái và lựa chọn action để đạt được mục tiêu.

2. Environment là gì?
- Environment là môi trường mà agent tương tác, cung cấp observation và reward sau mỗi action.

3. Observation khác action như thế nào?
- Observation là thông tin agent nhận được từ environment. Action là hành động agent lựa chọn để tác động lên environment.

4. Reward dùng để làm gì?
- Reward là tín hiệu phản hồi cho biết action của agent tốt hay xấu, được dùng để đánh giá và định hướng hành vi của agent.

5. Episode là gì?
- Episode là một lần tương tác hoàn chỉnh giữa agent và environment, bắt đầu từ reset() và kết thúc khi môi trường đạt trạng thái kết thúc.

6. Policy là gì?
- Policy là quy tắc hoặc chiến lược quyết định action nào agent sẽ chọn dựa trên observation hiện tại.

7. action_space thể hiện điều gì?
- action_space thể hiện tập hợp các action mà agent có thể thực hiện trong environment.


8. observation_space thể hiện điều gì?
- observation_space thể hiện không gian các observation mà agent có thể nhận được từ environment.


9. reset() được gọi khi nào?
- reset() được gọi khi bắt đầu một episode mới để đưa environment về trạng thái ban đầu và nhận observation đầu tiên.


10. step(action) thực hiện điều gì?
- step(action) đưa một action vào environment. Nó cập nhật trạng thái môi trường và trả về observation mới, reward, terminated, truncated và info.

11. terminated và truncated khác nhau như thế nào?
- terminated cho biết episode kết thúc do điều kiện kết thúc của environment. truncated cho biết episode bị dừng do giới hạn bên ngoài, chẳng hạn như giới hạn số bước.

12. Random policy có phải là một thuật toán tăng cường không? Giải thích.
- Không. Random policy chỉ chọn action một cách ngẫu nhiên và không học từ reward hoặc kinh nghiệm trong quá trình tương tác với environment.

13. Vì sao phải chạy nhiều episode thay vì chỉ một episode?
- Vì kết quả của mỗi episode có thể khác nhau do tính ngẫu nhiên. Chạy nhiều episode giúp đánh giá agent ổn định và đáng tin cậy hơn.

14. Vì sao cần random seed?
- Random seed giúp kiểm soát tính ngẫu nhiên, để có thể tái lập kết quả và dễ dàng so sánh các lần chạy thí nghiệm.

15. Rewward trung bình có ý nghĩa gì khi đánh giá agent?
- Reward trung bình cho biết mức reward mà agent đạt được trung bình qua nhiều episode, từ đó giúp đánh giá hiệu quả tổng thể của agent.
